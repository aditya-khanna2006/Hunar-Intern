# Assignment 2 - Statistics

## Objective
To calculate the mean height of UK males from historical data and visualize the trend using Python.


## Import Libraries

The required libraries are:
- Pandas for data handling
- Matplotlib for data visualization

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Creating the Dataset
The historical dataset contains the average height (in cm) of UK males from 1810 to 1980.

```python
data = {
    "Year": [1810, 1820, 1830, 1840, 1850, 1860, 1870, 1880,
             1900, 1910, 1920, 1930, 1940, 1950, 1960, 1970, 1980],
    "Height": [169.7, 169.1, 166.7, 166.5, 165.6, 166.6, 167.2,
               168.0, 169.4, 170.9, 171.0, 173.9, 174.9,
               176.0, 176.9, 177.1, 176.8]
}

df = pd.DataFrame(data)

df
```

## Calculating the Mean Height
The average height is calculated using the `mean()` function.

```python
mean_height = df["Height"].mean()

print(f"Mean Height: {mean_height:.2f} cm")
```

## Data Visualization
The following line chart shows the trend in average height of UK males over the years.

```python
plt.figure(figsize=(10,5))
plt.plot(df["Year"], df["Height"], marker='o')

plt.title("Average Height of UK Males (1810-1980)")
plt.xlabel("Year")
plt.ylabel("Height (cm)")
plt.grid(True)

plt.show()
```

## Conclusion

The mean height of UK males in the given dataset is **170.96 cm**. The line graph shows that the average height gradually increased from 1810 to 1980. This trend may reflect improvements in nutrition, healthcare, and overall living conditions over time.
